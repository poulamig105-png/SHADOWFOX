# Python Development Internship – Advanced Data Analysis

## Dataset: Iris

### Research Questions
1. How do sepal and petal measurements differ across species?
2. Which numerical features have the strongest relationships?
3. Can visualizations clearly separate the three species?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset('iris')
df.head()
```

```python
print('Shape:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nMissing values:')
print(df.isnull().sum())
print('\nSummary statistics:')
display(df.describe())
```

```python
# Data cleaning
df = df.drop_duplicates().dropna()
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

for feature in features:
    plt.figure(figsize=(7, 5))
    sns.boxplot(data=df, x='species', y=feature)
    plt.title(f'{feature.replace("_", " ").title()} by Species')
    plt.tight_layout()
    plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(df[features].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species')
plt.title('Petal Length vs Petal Width')
plt.tight_layout()
plt.show()
```

```python
# Pair plot
sns.pairplot(df, hue='species')
plt.show()
```

## Key Findings
- Petal measurements show clear differences among species.
- Petal length and petal width have a strong positive relationship.
- The visualizations help distinguish the species using their measurements.

## Conclusion
The Iris dataset demonstrates how exploratory data analysis and visualization can reveal patterns, relationships, and differences between groups in a dataset.
